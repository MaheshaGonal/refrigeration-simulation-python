```python
# CELL 2 — Imports
import matplotlib.pyplot as plt
import numpy as np
```

```python
# CELL 3 — The 20-day arc, by week and theme
weeks = [
    {"label": "Week 1\nDays 1-7", "theme": "Foundations:\nproperties, state points,\nP-H diagram, COP", "color": "#4FC3D9", "span": 7},
    {"label": "Week 2\nDays 8-14", "theme": "Component sizing:\ncap tube, condenser,\nreal compressor, climate", "color": "#E8A33D", "span": 7},
    {"label": "Week 3\nDays 15-20", "theme": "Patent-linked + capstone:\nPCM, defrost, BEE,\nmaster simulator", "color": "#D9764F", "span": 6},
]

fig, ax = plt.subplots(figsize=(12, 4))
start = 0
for w in weeks:
    ax.barh(0, w["span"], left=start, height=0.6, color=w["color"], alpha=0.85, edgecolor="white", linewidth=2)
    ax.text(start + w["span"]/2, 0, w["label"], ha="center", va="center", fontsize=11, fontweight="bold", color="white")
    ax.text(start + w["span"]/2, -0.85, w["theme"], ha="center", va="top", fontsize=9)
    start += w["span"]

ax.set_xlim(0, 20)
ax.set_ylim(-1.6, 0.8)
ax.set_yticks([])
ax.set_xticks(range(0, 21, 1))
ax.set_xlabel("Day")
ax.set_title("20 Days — Refrigeration Simulation Portfolio", fontweight="bold", fontsize=13)
for spine in ["top", "right", "left"]:
    ax.spines[spine].set_visible(False)
plt.tight_layout()
plt.savefig("day20_capstone_arc.png", dpi=150)
plt.show()

print(f"Total notebooks in repository: 20")
print(f"Refrigerants covered: R134a, R600a, R290")
print(f"Patents linked to simulation work: 2 of 3 (PCM pack; simultaneous cooling/defrosting)")
```
