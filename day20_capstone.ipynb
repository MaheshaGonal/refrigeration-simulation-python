```python
# CELL 2 — Imports
import matplotlib.pyplot as plt
import numpy as np
```

```python
# CELL 3 — The 20-day arc, by week and theme
import matplotlib.font_manager as fm
plt.rcParams['font.family'] = 'DejaVu Sans'

weeks = [
    {"label": "WEEK 1", "range": "Days 1\u20137", "theme": "Foundations\nproperties \u00b7 state points\nP-H diagram \u00b7 COP", "color": "#2E86AB", "span": 7},
    {"label": "WEEK 2", "range": "Days 8\u201314", "theme": "Component sizing\ncap tube \u00b7 condenser\nreal compressor \u00b7 climate", "color": "#F18F01", "span": 7},
    {"label": "WEEK 3", "range": "Days 15\u201320", "theme": "Patent-linked + capstone\nPCM \u00b7 defrost \u00b7 BEE\nmaster simulator", "color": "#C73E1D", "span": 6},
]

fig, ax = plt.subplots(figsize=(12, 6.3), dpi=150)
fig.patch.set_facecolor('#FAFAF8')
ax.set_facecolor('#FAFAF8')

start = 0
bar_h = 0.9
y_bar = 1.4
for w in weeks:
    ax.barh(y_bar, w["span"], left=start, height=bar_h, color=w["color"], edgecolor='#FAFAF8', linewidth=3, zorder=3)
    mid = start + w["span"]/2
    ax.text(mid, y_bar, w["label"], ha="center", va="center", fontsize=13, fontweight="bold", color="white", zorder=4)
    ax.text(mid, y_bar - bar_h/2 - 0.35, w["range"], ha="center", va="top", fontsize=10, color=w["color"], fontweight='bold', zorder=4)
    ax.text(mid, y_bar - bar_h/2 - 0.85, w["theme"], ha="center", va="top", fontsize=9, color="#333333", zorder=4, linespacing=1.6)
    start += w["span"]

for d in range(0, 21):
    ax.plot([d, d], [y_bar - bar_h/2 - 0.05, y_bar - bar_h/2], color='#999999', linewidth=0.8, zorder=2)

ax.set_xlim(-0.3, 20.3)
ax.set_ylim(-2.4, 3.0)
ax.axis('off')

ax.text(0, 2.6, "20", fontsize=54, fontweight='bold', color='#1A1A1A', ha='left', va='top')
ax.text(2.55, 2.75, "DAYS", fontsize=16, fontweight='bold', color='#666666', ha='left', va='top')
ax.text(2.55, 2.42, "Refrigeration Simulation Portfolio", fontsize=13, color='#444444', ha='left', va='top', style='italic')
ax.text(0, -2.15, "Mahesha Gonal  \u00b7  github.com/MaheshaGonal/refrigeration-simulation-python", fontsize=9.5, color='#888888', ha='left', va='bottom')

plt.tight_layout()
plt.savefig("day20_capstone_arc.png", dpi=150, facecolor=fig.get_facecolor())
plt.show()

print(f"Total notebooks in repository: 20")
print(f"Refrigerants covered: R134a, R600a, R290")
print(f"Patents linked to simulation work: 2 of 3 (PCM pack; simultaneous cooling/defrosting)")
```
